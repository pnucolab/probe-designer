```python
import pandas as pd
import matplotlib.pyplot as plt

# Read probe length benchmark results
data = []
with open('/home/abyot/projects/probeset/microbe_genome/benchmark_results.txt', 'r') as f:
    for line in f:
        line = line.strip()
        if not line or line.startswith('Probe') or line.startswith('-'):
            continue
        parts = line.split()
        if len(parts) == 2:
            data.append({'probe_length': int(parts[0]), 'time': float(parts[1])})

df = pd.DataFrame(data)
avg = df.groupby('probe_length')['time'].agg(['mean', 'std', 'count']).reset_index()
avg.columns = ['probe_length', 'mean_time', 'std_time', 'n_runs']

fig, ax = plt.subplots(figsize=(5, 5))

ax.bar(
    avg['probe_length'].astype(str),
    avg['mean_time'],
    yerr=avg['std_time'],
    capsize=5,
    color=['#3b82f6', '#10b981', '#f59e0b'],
    edgecolor='black',
    linewidth=0.8,
    width=0.85
)

for i, row in avg.iterrows():
    ax.text(
        i, row['mean_time'] + row['std_time'] + 0.1,
        f"{row['mean_time']:.2f}s",
        ha='center', va='bottom', fontsize=11, fontweight='bold'
    )

ax.set_xlabel('Probe Length (bp)', fontsize=12)
ax.set_ylabel('Execution Time (seconds)', fontsize=12)
ax.set_title(f'Average Execution Time vs Probe Length (n={avg["n_runs"].iloc[0]} runs)', fontsize=13)
ax.set_ylim(0, avg['mean_time'].max() + avg['std_time'].max() + 1)
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

plt.tight_layout()
plt.savefig('/home/abyot/projects/probeset/microbe_genome/benchmark_probe_length_plot.png', dpi=150)
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# Read benchmark results
data = []
with open('/home/abyot/projects/probeset/microbe_genome/benchmark_seqlength_results.txt', 'r') as f:
    for line in f:
        line = line.strip()
        if not line or line.startswith('Seq') or line.startswith('-'):
            continue
        parts = line.split()
        if len(parts) == 2:
            data.append({'seq_length_kb': int(parts[0]), 'time': float(parts[1])})

df = pd.DataFrame(data)
df['time_per_kb'] = df['time'] / df['seq_length_kb']
avg = df.groupby('seq_length_kb')['time_per_kb'].agg(['mean', 'std', 'count']).reset_index()
avg.columns = ['seq_length_kb', 'mean_time_per_kb', 'std_time_per_kb', 'n_runs']

print(avg.to_string(index=False))

# Plot
fig, ax = plt.subplots(figsize=(8, 5))

ax.errorbar(
    avg['seq_length_kb'],
    avg['mean_time_per_kb'],
    yerr=avg['std_time_per_kb'],
    fmt='-',
    color='#3b82f6',
    ecolor='#94a3b8',
    capsize=4,
    linewidth=2
)

for _, row in avg.iterrows():
    ax.annotate(
        f"{row['mean_time_per_kb']:.2f}",
        (row['seq_length_kb'], row['mean_time_per_kb']),
        textcoords="offset points", xytext=(0, 12),
        ha='center', fontsize=9, fontweight='bold'
    )

ax.set_yscale('log')
ax.set_xlabel('Sequence Length (kb)', fontsize=12)
ax.set_ylabel('Time per Kilobase (seconds/kb, log scale)', fontsize=12)
ax.set_title(f'Execution Time per Kilobase vs Sequence Length (n={avg["n_runs"].iloc[0]} runs)', fontsize=13)
ax.set_xticks(avg['seq_length_kb'])
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

plt.tight_layout()
plt.savefig('/home/abyot/projects/probeset/microbe_genome/benchmark_seqlength_plot.png', dpi=150)
plt.show()
```
